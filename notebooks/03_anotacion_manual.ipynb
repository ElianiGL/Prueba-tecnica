# Calificación en formato manual para validar al juez

1. Corre antes la evaluación (genera `resultados/anotacion_manual.csv`).
2. Para cada afirmación decide, **leyendo solo los fragmentos**, si está:
   - r = respaldada · n = no respaldada · c = contradicha · s = saltar
3. Al final se calcula el acuerdo con el juez.

````python
import os, sys
from pathlib import Path
RAIZ = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(RAIZ)); os.chdir(RAIZ)
````

````python
import pandas as pd
from IPython.display import Markdown, display, clear_output
from evaluacion.acuerdo_juez import RUTA_MUESTRA, calcular_acuerdo

m = pd.read_csv(RUTA_MUESTRA, dtype=str).fillna("")
MAPA = {"r": "respaldada", "n": "no_respaldada", "c": "contradicha"}
pendientes = m[m.etiqueta_humana == ""].index
print(f"{len(m) - len(pendientes)} / {len(m)} calificadas")
````

````python
for i in pendientes:
    f = m.loc[i]
    clear_output()
    display(Markdown(f"**{f['item']}** · {f['pregunta']}\n\n> **Afirmación:** {f['afirmacion']}\n\n<details open><summary>Fragmentos que vio el asistente</summary>\n\n```\n{f['fragmentos_contexto'][:12000]}\n```\n</details>"))
    x = input("r / n / c / s(altar) / q(uitar): ").strip().lower()
    if x == "q": break
    if x in MAPA:
        m.loc[i, "etiqueta_humana"] = MAPA[x]
        m.loc[i, "nota_humana"] = input("nota (opcional): ")
        m.to_csv(RUTA_MUESTRA, index=False)
print("Guardado.")
````

````python
res = calcular_acuerdo()
{k: v for k, v in (res or {}).items() if k not in ("desacuerdos",)}
````
